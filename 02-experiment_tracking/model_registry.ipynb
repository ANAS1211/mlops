```python
#interagir avec le serveur mlflow en utilisant la classe mlclient
from mlflow.tracking import MlflowClient
import mlflow

TRACKING_URI="sqlite:///mlflow.db"
client=MlflowClient(tracking_uri="TRACKING_URI")
```

```python
client.search_experiments()
```

```python
%pip list
mlflow.set_experiment("anas_experiment")
```

```python
experiment = client.get_experiment_by_name("anas_experiment")
print(f"Experiment ID: {experiment.experiment_id}")
mlflow.set_experiment("anas_experiment")
```

```python
client.create_experiment(name="anas_experiment")
```

```python
from mlflow.entities import ViewType

runs=client.search_runs(experiment_ids="1",
                   filter_string="metrics.rmse <6.8", 
                   run_view_type= ViewType.ACTIVE_ONLY,
                   max_results=5,
                     order_by= ["metrics.rmse ASC"])
```
